```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when, trim
```

```python
spark = SparkSession.builder \
    .appName("EinKaufPark Bronze Layer") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()
```

```python
bronze_path = os.path.join(os.getcwd(), "data", "bronze", "transactions")

bronze_df = spark.read.parquet(bronze_path)
```

```python
bronze_df.groupby("transaction_id").count().filter(col("count") > 1).show(20, truncate=False)
```

```python
bronze_df.filter(col("transaction_id") == "TXN-EKP-DE-010-C24A06849F43").show(5, truncate=False)
```
